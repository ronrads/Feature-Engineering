# 16. What is an Optimizer? What are different types of optimizers? Explain each with an example.
"""An optimizer is an algorithm that updates model parameters to minimize the loss function and improve accuracy.

Types include:

* **Gradient Descent**: Updates weights using full dataset; stable but slow.
* **SGD**: Updates with one sample at a time; faster but noisy.
* **Mini-Batch GD**: Uses small batches; balances speed and accuracy.
* **Momentum**: Accelerates convergence by considering past updates.
* **Adam**: Combines momentum and adaptive learning rates; widely used in deep learning.
